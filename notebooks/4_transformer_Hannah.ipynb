```python
import os
import pandas as pd
from email import policy
from email.parser import BytesParser

records = []

MAILDIR = "../data/maildir"

for sender in os.listdir(MAILDIR):

    sender_path = os.path.join(MAILDIR, sender)

    if not os.path.isdir(sender_path):
        continue

    for root, dirs, files in os.walk(sender_path):

        for file in files:

            filepath = os.path.join(root, file)

            try:
                with open(filepath, "rb") as f:
                    msg = BytesParser(policy=policy.default).parse(f)

                body = msg.get_body(preferencelist=('plain'))

                text = body.get_content() if body else ""

                records.append({
                    "sender": sender,
                    "text": text
                })

            except Exception:
                pass

df = pd.DataFrame(records)

print(df.shape)
df.head()
```

```python
df.to_csv("../data/enron_processed.csv", index=False)
```
